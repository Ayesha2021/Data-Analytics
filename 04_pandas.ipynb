```python
df = pd.DataFrame({
    'Company': ['GOOG', 'GOOG', 'MSFT', 'MSFT', 'FB', 'FB'],
    'Person': ['John', 'Sam', 'John', 'Jane', 'Mat', 'Nancy'],
    'Sales': [200, 150, 250, 300, 350, 400]
})

df
```

```python
df.groupby('Company').sum()
```

```python
dftemp =df.groupby('Company').sum()
dftemp
```

```python
company_sales = df.groupby('Company').sum()
type(company_sales)
company_sales.plot.bar()
plt.show()
```

```python
company_sales.plot.barh()

```

```python
df.groupby('Company').sum()['Sales'] #retuturns total sales per company as Series
```

```python
type(df.groupby('Company').sum()[['Sales']])
```

```python
company_sales=df.groupby('Company')['Sales'].sum()
company_sales.plot(kind='bar',color=['skyblue','lightgreen','lightsalmon'])
plt.title('Total Sales per Company')
plt.xlabel('Company')
plt.ylabel('Total Sales')
plt.grid(axis='y',linestyle='-',alpha=0.5) # alpha-0.5 is 50 percent opacity...1 meanns 100 percent opacity and dark
plt.show()

```

```python

plt.title('Sales Distribution per Company')
plt.ylabel('')
plt.show()
```

```python
df.groupby('Company').sum()['Sales']['GOOG']  # returns total sales for GOOG company
```

Total Sales per Person

```python
sales_person=df.groupby('Person').sum()['Sales']  # returns total sales per person as Series
sales_person
```

```python
sales_person.plot(kind='bar',color=['red','green','blue','orange','yellow'])
plt.title('Total Sales Per Person')
plt.xlabel('Person')
plt.ylabel('Totoal Sales')
plt.grid(axis='y',linestyle='-',alpha=0.5)  # add grid lines to y-axis
plt.show()
```

Count of Sales per Person

```python
count_sale_person= df.groupby('Person').count()['Sales']
count_sale_person.plot(kind='bar',color=['red','green','blue','orange','yellow'])
plt.title('Count of Sales per Person')
plt.xlabel('Person')
plt.ylabel('Count of Sales')
plt.grid(axis='y',linestyle='-',alpha=0.5)
plt.show()
```

```python
count_sale_person1= df.groupby('Person').count()['Sales']
count_sale_person1.plot(kind='barh',color=['red','green','blue','orange','yellow'])
plt.title('Count of Sales per Person')
plt.xlabel('Count of Sales')
plt.ylabel('Person')
plt.grid(axis='y',linestyle='-',alpha=0.5)
plt.show()
```
