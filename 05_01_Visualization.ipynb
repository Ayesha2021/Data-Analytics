```python

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
pd.set_option('display.max_rows', 20)  # Show 2o rows in DF instead of defafult 10.
%matplotlib inline  
```

```python
x=np.linspace(0,5,10)  # Create an array of 10 points between 0 and 5
print(x)
```

```python

y = x**2
print(y)
```

```python
plt.title('Numbers and Squares')
plt.xlabel('Numbers')
plt.ylabel('Squares')
plt.plot(x,y,'rs--')# valid values for 3rd parameters are ro bs 1 

plt.legend(['Numbers','Squares'])
plt.grid(True)
#plt.xticks(np.arange(0, 6, 1))  # Set x-ticks from 0 to 5
#plt.yticks(np.arange(0, 26, 5))
plt.savefig('plot1.png', dpi=300, bbox_inches='tight')  # Save the plot as a PNG file and save before showing .
plt.show()
```

Working with Multiple Plots

```python
plt.subplot(1,2,1) #1 row, 2 columns, first subplot
plt.plot(x,y,'r')

plt.subplot(1,2,2) #1 row, 2 columns, second subplot
plt.plot(y,x,'g')
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'r*-')
plt.subplot(1,2,2)
plt.plot(y,x,'go--')
plt.show()
```

###Object  Oriented Model of Accessing plot

```python
figure1 =plt.figure()
axes = figure1.add_axes([0.1,0.1,0.8,0.8]) # [left, bottom, width, height] ,canvas dimentions
axes.plot(x,y,'r*-')
axes.set_xlabel('Numbers')
axes.set_ylabel('Squares')
axes.set_title('Numbers and Squares')
figure1.show()
```

```python
figure1 = plt.figure()
axes = figure1.add_axes([0.1, 0.1, 0.8, 0.8])
axes.plot(x, y, 'ro--')
axes.set_title('Numbers and Squares')
axes.set_xlabel('Numbers')
axes.set_ylabel('Squares')
axes.grid(True) # try without this line
figure1.show()
```

```python
figure2 =plt.figure() 
axes1=figure2.add_axes([0.1,0.1,0.8,0.8]) # [left, bottom, width, height] ,canvas dimentions
axes1.set_title('Outer Numbers and Squares')
axes1.set_xlabel('Outer Numbers')
axes1.set_ylabel('Outer Squares')
axes1.plot(x,y,'r*-')

axes2 =figure2.add_axes([0.2,0.5,0.4,0.3])
axes2.set_title('Inner Numbers and Squares')
axes2.set_xlabel('Inner Numbers')
axes2.set_ylabel('Inner Squares')
axes2.plot(y,x,'go--')

figure2.show()
```

```python
fig ,axes = plt.subplots(1,2)
colors=['ro-','gs--']

for i in range(len(axes)):
    axes[i].set_xlabel('Numbers')
    axes[i].set_ylabel('Squares')
    axes[i].set_title('Numbers and Squares')
    axes[i].plot(x,y,colors[i])
```

Dual Plots

```python
fig = plt.figure()

axes = fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,x**2, color='r',label='Numbers and Squares',linestyle='--',marker='o')
axes.plot(x,x**3, color='g',label='Numbers and Cubes',linestyle='--',marker='s')
axes.legend(loc=0)
axes.legend(loc=9)
axes.set_xlabel('Numbers')
axes.set_ylabel('Squares ')
axes.set_title('Numbers, Squares and Cubes')
axes.grid(True)
fig.show()
```
